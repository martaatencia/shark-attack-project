```python
import re

# 1. First, try to parse dates using pandas to_datetime (handles standard formats like 2025-06-11)
parsed_dates = pd.to_datetime(refined_dataset['date'], errors='coerce')

# Extract month name from successfully parsed datetime objects
refined_dataset['month'] = parsed_dates.dt.month_name()

# 2. For rows where pandas couldn't parse it (e.g., "20th October", historical dates), use regex fallback
months_list = ['January', 'February', 'March', 'April', 'May', 'June',
               'July', 'August', 'September', 'October', 'November', 'December',
               'Jan', 'Feb', 'Mar', 'Apr', 'May', 'Jun', 'Jul', 'Aug', 'Sep', 'Oct', 'Nov', 'Dec'
               ]
pattern = '|'.join(months_list)

# Fill missing months using regex extraction on the original text
mask_missing = refined_dataset['month'].isnull()
refined_dataset.loc[mask_missing, 'month'] = (
    refined_dataset.loc[mask_missing, 'date']
    .str.extract(f'({pattern})', flags=re.IGNORECASE)[0]
)

# Check how many rows are still without a month
missing_months = refined_dataset['month'].isnull().sum()
print(f"Number of rows without a month: {missing_months}")

# Show some examples of what is still missing (truly historical or irregular entries)
print("\nExamples of remaining missing dates:")
print(refined_dataset[refined_dataset['month'].isnull()]['date'].head(10))

# Check the distribution of the newly populated month column
print("\nMonth distribution:")
print(refined_dataset['month'].value_counts())
```

```python
# Map short month names to full names to clean up the distribution
month_mapping = {
    'Jan': 'January', 'Feb': 'February', 'Mar': 'March', 'Apr': 'April',
    'May': 'May', 'Jun': 'June', 'Jul': 'July', 'Aug': 'August',
    'Sep': 'September', 'Oct': 'October', 'Nov': 'November', 'Dec': 'December'
}

# Apply mapping (we can capitalize the first 3 letters to match the keys safely)
refined_dataset['month'] = refined_dataset['month'].replace(month_mapping)

# Check the final, clean month distribution
print("Cleaned and unified Month distribution:")
print(refined_dataset['month'].value_counts())
```

```python
# Check in how many rows we did not find a month.
missing_months = refined_dataset['month'].isnull().sum()
print(f"Number of rows without a month: {missing_months}")

# Show a few examples where the month became NaN.
print(refined_dataset[refined_dataset['month'].isnull()]['date'].head(10))

```
